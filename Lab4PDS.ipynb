```python
import pandas as pd

student_data = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Math": [85, 90, None, 60, 78],
    "Science": [88, None, 92, 70, 80],
    "English": [82, 76, 89, None, 95]
})

student_data.fillna(student_data.mean(numeric_only=True), inplace=True)

high_achievers = student_data[
    (student_data["Math"] > 75) &
    (student_data["Science"] > 75) &
    (student_data["English"] > 75)
]

print("=== High Achievers (scored above 75 in all subjects) ===")
print(high_achievers.to_string(index=False))
```

```python
import pandas as pd

patient_data = pd.DataFrame({
    "Patient_ID": [201, 202, 203, 204, 205, 206],
    "Name": ["Mary", "John", "Linda", "Robert", "Patricia", "Susan"],
    "Age": [65, None, 72, 45, 58, None],
    "Gender": ["Female", "Male", "Female", "Male", "Female", "Female"],
    "Diagnosis": ["Diabetes", "Flu", "Diabetes", "Cancer", "Diabetes", "Diabetes"],
    "Admission_Date": ["2025-08-01", "2025-08-03", "2025-08-05", "2025-08-06", "2025-08-07", "2025-08-09"]
})

patient_data["Age"] = patient_data["Age"].fillna(patient_data["Age"].median())

elderly_female_diabetes = patient_data[
    (patient_data["Gender"].str.lower() == "female") &
    (patient_data["Age"] > 50) &
    (patient_data["Diagnosis"].str.lower() == "diabetes")
]

print("=== Elderly Female Patients Diagnosed with Diabetes ===")
print(elderly_female_diabetes.to_string(index=False))
```
