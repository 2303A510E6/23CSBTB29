```python
import pandas as pd

data = {'Patient_ID': [1, 2, 3, 4, 5, 6],
        'Age': [25, 34, -5, 45, 130, 60]}
df = pd.DataFrame(data)

print("Before Cleaning:\n", df)

outliers = df[(df['Age'] < 0) | (df['Age'] > 120)]
print("\nDetected Outliers:\n", outliers)

cleaned_df = df[(df['Age'] >= 0) & (df['Age'] <= 120)]

print("\nAfter Cleaning:\n", cleaned_df)
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

data = {'Customer_ID': [101, 102, 103, 104, 105],
        'Income': [20000, 35000, 50000, 75000, 100000]}
df = pd.DataFrame(data)

print("Original Data:\n", df)

scaler_minmax = MinMaxScaler()
df['Income_MinMax'] = scaler_minmax.fit_transform(df[['Income']])

scaler_standard = StandardScaler()
df['Income_Standardized'] = scaler_standard.fit_transform(df[['Income']])

print("\nAfter Scaling:\n", df)
```
