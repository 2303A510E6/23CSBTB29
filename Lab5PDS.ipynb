```python
import pandas as pd

customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'city': ['Delhi', 'Mumbai', 'Delhi', 'Bangalore']
})

transactions = pd.DataFrame({
    'transaction_id': [101, 102, 103, 104, 105, 106],
    'customer_id': [1, 2, 1, 3, 4, 2],
    'amount': [500, 1200, 700, 1500, 800, 600]
})

print("Customers:\n", customers)
print("\nTransactions:\n", transactions)

merged = pd.merge(transactions, customers, on='customer_id')
print("\nMerged Dataset:\n", merged)

customer_spend = merged.groupby(['customer_id', 'name'])['amount'].sum().reset_index()
print("\nTotal Spend per Customer:\n", customer_spend)

city_spend = merged.groupby('city')['amount'].sum().reset_index()
print("\nCity-wise Spending:\n", city_spend)

top_spender = customer_spend.loc[customer_spend['amount'].idxmax()]
print("\nTop Spender:\n", top_spender)
```

```python
import pandas as pd

patients = pd.DataFrame({
    'patient_id': [1, 2, 3, 4],
    'name': ['Ravi', 'Meena', 'Suresh', 'Priya'],
    'age': [30, 45, 28, 60]
})

appointments = pd.DataFrame({
    'appointment_id': [201, 202, 203, 204, 205, 206, 207],
    'patient_id': [1, 2, 1, 3, 2, 4, 1],
    'doctor': ['Dr. A', 'Dr. B', 'Dr. A', 'Dr. C', 'Dr. B', 'Dr. C', 'Dr. A']
})

print("Patients:\n", patients)
print("\nAppointments:\n", appointments)

merged = pd.merge(appointments, patients, on='patient_id')
print("\nMerged Dataset:\n", merged)

patient_visits = merged.groupby(['patient_id', 'name'])['appointment_id'].count().reset_index(name='visits')
print("\nAppointments per Patient:\n", patient_visits)

doctor_workload = merged.groupby('doctor')['appointment_id'].count().reset_index(name='appointments')
print("\nDoctor Workload:\n", doctor_workload)

max_patient = patient_visits.loc[patient_visits['visits'].idxmax()]
print("\nPatient with Maximum Visits:\n", max_patient)
```
